# Task
Compare the performance of ELMo and BERT embeddings for sentence classification using Naive Bayes, Logistic Regression, and Support Vector Machine classifiers. The task involves preparing the provided `sentences` and `labels` data, generating embeddings with both ELMo and BERT models, training and evaluating each classifier with both types of embeddings, analyzing the impact of feature size, and investigating how well ELMo and BERT handle ambiguous sentences like "The bat is flying" and "He hit the ball with a bat". Finally, summarize all experimental findings, including performance metrics, feature size observations, and insights from the ambiguous sentence test.

## Prepare Data for Modeling

### Subtask:
Split the 'sentences' and 'labels' into training and testing sets to prepare for model training and evaluation. Ensure a consistent split for fair comparison between models.


**Reasoning**:
To split the data into training and testing sets, I need to import `train_test_split` from `sklearn.model_selection` and then apply it to the `sentences` and `labels` using the specified parameters.



````python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(sentences, labels, test_size=0.2, random_state=42)

print(f"Training set size: {len(X_train)} samples")
print(f"Testing set size: {len(X_test)} samples")
print(f"First 5 training sentences: {X_train[:5]}")
print(f"First 5 training labels: {y_train[:5]}")
````

## Load ELMo Model and Generate Embeddings

### Subtask:
Load the pre-trained ELMo model and use it to generate contextualized word embeddings for all sentences in both the training and testing sets. This step will require handling potential library installations if ELMo is not readily available.


**Reasoning**:
The first instruction is to install `tensorflow_text` and `tensorflow_hub` if they are not already installed. I will use a code block with `pip install` commands to ensure these libraries are available.



````python
get_ipython().system('pip install tensorflow_text tensorflow_hub')
print("Tensorflow Text and Tensorflow Hub libraries are installed.")
````

**Reasoning**:
Now that the necessary libraries are installed, the next step is to import `tensorflow_hub` and `tensorflow_text`, load the ELMo model, and then generate the ELMo embeddings for both the training and testing sets as per instructions 2, 3 and 4 of the subtask.



````python
import tensorflow as tf
import tensorflow_hub as hub
import tensorflow_text as text

# Load the pre-trained ELMo model
elmo_model = hub.load("https://tfhub.dev/google/elmo/3")

# Generate ELMo word embeddings for training and testing sentences
# The 'default' signature processes the sentences and returns a dictionary.
# We access the 'elmo' key for the contextualized word representations.
# Then, we average these word embeddings across the sequence length to get sentence embeddings.
elmo_output_train = elmo_model.signatures["default"](tf.constant(X_train))
elmo_word_embeddings_train = elmo_output_train["elmo"]
elmo_train_embeddings = tf.reduce_mean(elmo_word_embeddings_train, axis=1)

elmo_output_test = elmo_model.signatures["default"](tf.constant(X_test))
elmo_word_embeddings_test = elmo_output_test["elmo"]
elmo_test_embeddings = tf.reduce_mean(elmo_word_embeddings_test, axis=1)

print("ELMo embeddings generated for training and testing sets.")
print(f"Shape of ELMo training embeddings: {elmo_train_embeddings.shape}")
print(f"Shape of ELMo testing embeddings: {elmo_test_embeddings.shape}")
````

## Apply Mean Pooling and Train Naive Bayes with ELMo

### Subtask:
Perform mean pooling on the ELMo word embeddings to obtain fixed-size sentence embeddings. Then, train a Naive Bayes classifier using these ELMo-based sentence embeddings from the training set.


**Reasoning**:
The subtask requires training a Naive Bayes classifier. I will import the `GaussianNB` classifier, instantiate it, and then train the model using the previously generated `elmo_train_embeddings` and `y_train`.



````python
from sklearn.naive_bayes import GaussianNB

# Instantiate the Gaussian Naive Bayes classifier
naive_bayes_elmo_model = GaussianNB()

# Train the Naive Bayes model using ELMo embeddings for training data
naive_bayes_elmo_model.fit(elmo_train_embeddings.numpy(), y_train)

print("Naive Bayes classifier trained successfully using ELMo embeddings.")
````

## Evaluate ELMo + Naive Bayes

### Subtask:
Evaluate the performance of the trained ELMo + Naive Bayes model on the test set. Calculate and store evaluation metrics such as accuracy, and also record the feature size of the ELMo embeddings for later comparison.


**Reasoning**:
To evaluate the ELMo + Naive Bayes model, I will import `accuracy_score`, make predictions on the test set, calculate and print the accuracy, and then extract and print the feature size of the ELMo embeddings as instructed.



````python
from sklearn.metrics import accuracy_score

# Make predictions on the ELMo test embeddings
elmo_nb_predictions = naive_bayes_elmo_model.predict(elmo_test_embeddings.numpy())

# Calculate accuracy
elmo_nb_accuracy = accuracy_score(y_test, elmo_nb_predictions)
print(f"ELMo + Naive Bayes Accuracy: {elmo_nb_accuracy:.4f}")

# Get and store the feature size of ELMo embeddings
elmo_feature_size = elmo_train_embeddings.shape[1]
print(f"ELMo Embedding Feature Size: {elmo_feature_size}")
````

## Load BERT Tokenizer and Model

### Subtask:
Load a pre-trained BERT tokenizer and model suitable for generating sentence embeddings from the `transformers` library. I will use a general-purpose BERT model like `bert-base-uncased`.


**Reasoning**:
The first step in loading the BERT tokenizer and model is to install the `transformers` library, as specified in the instructions. I will use a `code_block` to execute the `pip install` command.



````python
get_ipython().system('pip install transformers')
print("Transformers library installed.")
````

**Reasoning**:
Now that the `transformers` library is installed, I will proceed to import `BertTokenizer` and `BertModel` and then load the pre-trained 'bert-base-uncased' tokenizer and model as specified in the subtask instructions.



````python
from transformers import BertTokenizer, BertModel
import torch

# Load the pre-trained BertTokenizer
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# Load the pre-trained BertModel
model = BertModel.from_pretrained('bert-base-uncased')

print("BERT Tokenizer and Model loaded successfully using 'bert-base-uncased'.")
````

**Reasoning**:
Now that the BERT tokenizer and model are loaded, the next step is to tokenize the training and testing sentences and generate their BERT embeddings, which is crucial for preparing the data for subsequent classifier training.



````python
def get_bert_embeddings(sentences, tokenizer, model):
    encoded_input = tokenizer(sentences, padding=True, truncation=True, return_tensors='pt')
    with torch.no_grad():
        model_output = model(**encoded_input)
    # Get the embeddings of the [CLS] token, which often serves as a sentence embedding
    # or mean-pool the last hidden states
    # For simplicity, let's use the [CLS] token embedding here
    # The last_hidden_state is (batch_size, sequence_length, hidden_size)
    # The [CLS] token is at index 0
    embeddings = model_output.last_hidden_state[:, 0, :]
    return embeddings.numpy()

# Generate BERT embeddings for training and testing sets
bert_train_embeddings = get_bert_embeddings(X_train, tokenizer, model)
bert_test_embeddings = get_bert_embeddings(X_test, tokenizer, model)

print("BERT embeddings generated for training and testing sets.")
print(f"Shape of BERT training embeddings: {bert_train_embeddings.shape}")
print(f"Shape of BERT testing embeddings: {bert_test_embeddings.shape}")
````

## Train and Evaluate BERT + Naive Bayes

### Subtask:
Train a Naive Bayes classifier using the generated BERT embeddings from the training set and evaluate its performance on the test set. Calculate and record the accuracy and BERT embedding feature size.


**Reasoning**:
The subtask requires training and evaluating a Naive Bayes classifier with BERT embeddings. I will import the necessary classifier and metric, train the model, make predictions, calculate accuracy, and then extract the feature size, all within a single code block.



````python
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

# Instantiate the Gaussian Naive Bayes classifier
naive_bayes_bert_model = GaussianNB()

# Train the Naive Bayes model using BERT embeddings for training data
naive_bayes_bert_model.fit(bert_train_embeddings, y_train)

print("Naive Bayes classifier trained successfully using BERT embeddings.")

# Make predictions on the BERT test embeddings
bert_nb_predictions = naive_bayes_bert_model.predict(bert_test_embeddings)

# Calculate accuracy
bert_nb_accuracy = accuracy_score(y_test, bert_nb_predictions)
print(f"BERT + Naive Bayes Accuracy: {bert_nb_accuracy:.4f}")

# Get and store the feature size of BERT embeddings
bert_feature_size = bert_train_embeddings.shape[1]
print(f"BERT Embedding Feature Size: {bert_feature_size}")
````

Train a Logistic Regression classifier using the ELMo embeddings from the training set and evaluate its performance on the test set. Calculate and record the accuracy.
